#### Task 5: 

Complete the function `isPerfectNumber` which takes a positive integer `n` and returns whether `n` is a perfect number or not.
- Note that: A perfect number is a positive integer that is equal to the sum of its positive proper divisors (all positive divisors excluding the number itself). For example, $6$ is a perfect number because its proper divisors are $1$, $2$, and $3$, and $1 + 2 + 3 = 6$. The number $1$ is not a perfect number.

```python
def isPerfectNumber(n):
    # YOUR CODE HERE
    # 判断传入的n是否为完数
    sum = 0
    for i in range(1, n):
        if n % i == 0:
            sum += i
    if sum == n:
        return True
    else:
        return False
```

```python
assert isPerfectNumber(6) == True
assert isPerfectNumber(28) == True
assert isPerfectNumber(12) == False
assert isPerfectNumber(1) == False
```
