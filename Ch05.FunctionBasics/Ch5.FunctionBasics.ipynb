## 1. Baiscs of function

### Syntax of Function

The syntax of function is

```python
def function_name(parameters):
	"""docstring"""
	statement(s)
```

Below is a list of key points to define a Python function:
* The keyword `def` is used to define a function in Python.
* To call a function, use the `function_name` followed by parenthesis.
* `Arguments(input parameters)` are specified after the function name, inside the parentheses. You can add as many arguments as you want, just separate them with a comma.
* Lines and indentation are used to indicate the function block.
* Python `return` statement to let a function return a value.
* Python function has an optional documentation string (docstring) which describes what the function does.

### Function Example

A function is not executed until it is called. The following code defines and calls a function which does not return a value.

````python
# function greet(name) prints greeting message to name
def greet(name):
    """This function greets to
    the person passed in as
    parameter"""
    print("Hello, " + name + ". Nice to meet you!")


# function greet(name) is called
greet(input("What is your name? "))
````

And the following code defines and calls a function which returns a string.

````python
# function greet(name) returns a greeting message to name
def greet(name):
    """This function returns a
    greeting message to the person
     passed in as parameter"""
    return "Hello, " + name + ". Nice to meet you!"


# function greet(name) is called
msg = greet(input("What is your name? "))
print(msg)
````

### Docstring

The first string after the function header is called the *docstring* and is short for documentation string. It is used to explain in brief, what a function does.

Although optional, documentation is a good programming practice. In both of the two examples above, we have docstrings immediately below the function header. We generally use triple quotes so that docstring can extend up to multiple lines. This string is available to us as `__doc__` attribute of the function.

For example:

Try running the following into the Python shell to see the output.

````python
import numpy
def greet(name):
    """This function returns a
    greeting message to the person
     passed in as parameter"""
    return "Hello, " + name + ". Nice to meet you!"


print(greet.__doc__)
help(greet)
````

If there is no expression in the statement or the return statement itself is not present inside a function, then the function will return the `None` object. For example, the following code

## 2. Recursion
Python also accepts function recursion, which means a defined function can call itself.

Recursion is a common mathematical and programming concept. `It means that a function calls itself.` This has the benefit of meaning that you can loop through data to reach a result.

The developer should be very careful with recursion as it can be quite easy to slip into writing a function which never terminates, or one that uses excess amounts of memory or processor power. However, when written correctly recursion can be a very efficient and mathematically-elegant approach to programming.

In this example, `sum_number(n)` is a function that we have defined to call itself ("recurse"). We use the `n` variable as the data, which increase 1 every time we recurse. The recursion ends when the condition is no longer valid.

````python
def sum_number(n):
    total = 0
    for i in range(1, n+1):
        total += i
    return total

print(sum_number(100))
````

````python
def sum_number(n):
    if n <= 0:
        return 0
    return n+sum_number(n-1)

print(sum_number(100))
````

## 3. Python Lambda

A lambda function is a small anonymous function, which can take any number of arguments, but can only have one expression.

The syntax of function Lambda is

```Python
lambda arguments : expression
```

The expression is executed and the result is returned.

For Example, we want to add 10 to argument a, and return the result:

````python
x = lambda a : a + 10
print(x(5))
````

Lambda functions can take any number of arguments, For example:

````python
x = lambda a, b, c : a + b + c
print(x(5, 6, 2))
````

The power of lambda is better shown when you use them as an anonymous function inside another function.

````python
# Program to filter out only the even items from a list
my_list = [1, 5, 4, 6, 8, 11, 3, 12]
new_list = list(filter(lambda x: (x%2 == 0) , my_list))

print(new_list)
````
