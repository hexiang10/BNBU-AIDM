#### Task 1:

In the code cell below, complete the function `minorityDifference` which accepts two sets, `A` and `B` as arguments. The function returns True if the symmetric difference of A and B (elements present in A or B, but not both) is strictly smaller in size than the intersection of A and B (elements present in both). Otherwise, return False.

```python
def minorityDifference(A, B):
    # YOUR CODE HERE
    return len(A ^ B) < len(A & B)
```

```python
assert minorityDifference({1, 2, 3, 4}, {2, 3, 4}) == True
assert minorityDifference({1, 2, 3}, {2, 3, 4}) == False
assert minorityDifference({1, 2}, {3, 4}) == False
assert minorityDifference(set(), set()) == False
```

In the code cell below, complete the function `commonWords` which accepts a list of strings `strings` as argument. It returns a set containing all distinct words that appear in every string in the list. If strings is empty, return an empty set.

```python
def commonWords(strings):
    # If the input list is empty, return an empty set
    if not strings:
        return set()
    
    # Split the first string into words and create an initial set
    common = set(strings[0].split())
    
    # Intersect with every other string's word set
    for s in strings[1:]:
        common &= set(s.split())
    
    return common


commonWords(["hello world"])

    
    
```

```python
strings = ["hello world"]
str_set  = [set(s.split()) for s in strings]
result = str_set[0]
for s in range(len(str_set)):
    result = str_set[s] & result
print(result)
```

```python
assert commonWords(["a deal is a deal", "deal with it", "a good deal"]) == {"deal"}
assert commonWords(["apple banana", "banana cherry", "banana pie"]) == {"banana"}
assert commonWords(["hello world", "python code"]) == set()
assert commonWords([]) == set()
```

#### Task 3:

In the code cell below, complete the function `uniqueToFirst` which accepts a string `str1` and a list of strings `strings` as arguements. It returns a set containing all distinct words that appear in the first string `str1` but do not appear in any of the subsequent strings in the list of `strings`. If either of `str1` or `strings` is empty, return an empty set.

```python
def uniqueToString(str1, strings):
    # YOUR CODE HERE
    if not str1 or not strings:
        return set()
    result = set(str1.split())
    for s in strings:
        result-=set(s.split())
    return result

str1 = "cat sat on mat"
strings = []
uniqueToString(str1, strings)
```

#### Task 4:

In the code cell below, complete the function `filterWords` which accepts a list of strings `words` and a set of characters `chars`. The function returns a new list containing only the words whose unique characters all from the `chars` while preserving the original order of words.

*Challenge*: Can you complete the task with one line of code?

```python
def filterWords(words, chars):
    # YOUR CODE HERE
    return [word for word in words if set(word) <= chars]

words = ["apple", "banana"]
chars = {'a', 'p', 'l', 'e'}
filterWords(words, chars)
```

```python
words = ["cat", "hat", "at"]
chars = {'a', 'c', 't', 'h'}
assert filterWords(words, chars) == ["cat", "hat", "at"]
```

In the code cell below, complete the function `isValidPartition` which takes a set `universe` and a list of sets `subsets` as argument. The function returns True if subsets forms a valid partition of universe, and False otherwise. A valid partition must satisfy three conditions:

- Every set in `subsets` must be non-empty.
- The union of all sets in `subsets` must be exactly equal to `universe`.
- Any two sets in `subsets` do not share common elements.

For example:

- Given universe = {1, 2, 3, 4, 5} and subsets =[{1, 2}, {3, 4}, {5}], it should return Ture.

- Given universe = {1, 2, 3, 4} and subsets =[{1, 2}, {2, 3, 4}], it should return False.

```python
def isValidPartition(universe, subsets):
    # Condition 1: Every set in subsets must be non-empty
    for s in subsets:
        if not s:  # empty set
            return False
    
    # Condition 3: Any two sets must be disjoint
    n = len(subsets)
    for i in range(n):
        for j in range(i + 1, n):
            if subsets[i].intersection(subsets[j]):
                return False
    
    # Condition 2: Union of all subsets must equal universe
    union_set = set()
    for s in subsets:
        union_set.update(s)
    
    if union_set != universe:
        return False
    
    # All conditions satisfied
    return True
```

```python
universe = {1, 2, 3, 4}
subsets =[{1, 2}, {3, 4}]
assert isValidPartition(universe, subsets) == True
```

```python
universe = {1, 2, 3}
subsets =[{1, 2}, {2, 3}]
assert isValidPartition(universe, subsets) == False
```

```python
universe = set()
subsets =[]
assert isValidPartition(universe, subsets) == True
```
