A dictionary is like an address-book where you can find the address or contact details of a person by knowing only their name, i.e. we associate **keys** (name) with **values** (details).


## 1. Ceate Dictionary

To create a dictionary, you may place all the dictionary items inside a pair of curly brackets ({}), separated by commas. Follows are some notes about the declaration of a dictionary.

* Every item is defined by a *key* and a *value*, separated by a colon (:).
* You can use **only immutable objects (like strings) for the keys**.
* The keys must be **unique**.
* You can use either immutable or mutable objects for the values of the dictionary.

```python
# This creates an empty dictionary
dic = {}

# This creates a dictionary, the names are the "keys", and the emails are the "values".
dic = {'Alice':'alice@uic.edu.hk', 'Bob':'bob@uic.edu.hk'}

# This creates a dictionary with mixed keys
my_dict = {'name': 'John', 1: [1, 2, 3]}#

# This creates a dictionary from a sequence having each item as a pair
my_dict = dict([(1,'apple'), (2,'pineapple')])
print(my_dict)
```

```python
# In this example, the two pairs have the same key 'Alice'
# The second pair is not added into the dictionary as a new pair
# Instead, it is used to update the first pair, so there will be only one pair in the dictionary
dic = {'Alice':'alice@uic.edu.hk', 'Alice':'ali@uic.edu.cn'}
print(dic)
```

```python
# This raises a TypeError, since key ['Alice'] is mutable.
dic = {['Alice']:'alice@uic.edu.hk', 'Bob':'bob@uic.edu.hk'}
```

## 2. Access Values in Dictionary

There are two ways to get the values in a dictionary.

* Use the key as the index. If the key does not exist in the dictionary, a KeyError will be raised.

```python
dic = {'Alice':'alice@uic.edu.hk', 'Bob':'bob@uic.edu.hk'}

# This prints the value of key 'Bob'
print(dic['Bob'])

# Key 'Cici' is not in the dictionary, this causes a KeyError
print(dic['Cici'])
```

* Use the `get()` method. If the key does not exist in the dictionary, a default value will be returned.

```python
# Key 'Cici' is not in the dictionary, so None will be returned by default
print(dic.get('Cici'))

# You can set the default value for the non-existant keys
print(dic.get('Cici', 'Mail not found'))
```

## 3. Change or Add Dictionary Elements

Dictionaries are *mutable*. We can add new items or change the value of existing items using an assignment operator.

If the key is already present, then the existing value gets updated. In case the key is not present, a new `key: value` pair is added to the dictionary.

```python
# Changing and adding Dictionary Elements
dic = {'Alice':'alice@uic.edu.hk', 'Bob':'bob@uic.edu.hk'}

# update value
dic['Alice'] = 'alice@uic.edu.cn'
print(dic)

# add item
dic['Cici'] = 'cici@uic.edu.cn'
print(dic)
```

## 4. Remove Elements from Dictionary
There are at least 4 ways that we can remove elements from a dictionary.

* Use the `pop()` method. Given a `key`, it removes the element with the key and returns its `value`.
* Use the `popitem()` method to remove and return a random element from the dictionary.
* Use the `clear()` method to empty the dictionary.
* Use the `del` keyword to remove individual items or the entire dictionary itself.

```python
dic = {'Alice':'alice@uic.edu.hk', 'Bob':'bob@uic.edu.hk', 'Cici':'cici@uic.edu.hk', 'Dan':'dan@uic.edu.hk'}

# This removes the element with key 'Alice'
print(dic.pop('Alice'))
print(dic)

# This removes the last element insert to the dictionary.
# Before version Python3.7, it work as removing a random element.
print(dic.popitem())
print(dic)

# This removes all the elements
# The dictionary is empty but it is still there
dic.clear()
print(dic)

# This destroys the dictionary entirely
# After del, trying to access dic will raise a NameError
del dic
print(dic)
```

##  5. Dictionary Methods

Methods that are available with a dictionary are listed below. You may check [here](https://www.programiz.com/python-programming/methods/dictionary) for further details.

|Method|Description|
| ---  | --- |
|clear()|Removes all items from the dictionary.|
|copy()|Returns a shallow copy of the dictionary.|
|fromkeys(seq[, v])|Returns a new dictionary with keys from `seq` and value equal to `v` (defaults to `None`).|
|get(key[,d])|Returns the value of the `key`. If the `key` does not exist, returns `d` (defaults to `None`).|
|items()|Return a new object of the dictionary's items in (key, value) format.|
|keys()|Returns a new object of the dictionary's keys.|
|pop(key[,d])|Removes the item with the `key` and returns its value or `d` if `key` is not found. If `d` is not provided and the `key` is not found, it raises `KeyError`.|
|popitem()|Removes and returns an arbitrary item *(key, value)*. Raises `KeyError` if the dictionary is empty.|
|setdefault(key[,d])|Returns the corresponding value if the `key` is in the dictionary. If not, inserts the `key` with a value of `d` and returns `d` (defaults to `None`).|
|update([other])|Updates the dictionary with the key/value pairs from `other`, overwriting existing keys.|
|values()|Returns a new object of the dictionary's values|

## 7. Dictionary Comprehensions

Dictionary comprehension is an elegant and concise way to create a new dictionary from an iterable in Python. Its syntax is similar to that of a list comprehension.

### Advantage of Using Dictionary Comprehension

Using dictionary comprehension in our code can shorten the lines of code while keeping the logic intact.


A very common task in Python is to take some kind of sequential data and then turn it into a dictionary. Say, for example, that we wanted to take a list of strings and then create a dictionary mapping the strings to their lengths.

**Example:** Record the `province name: name length` pair for the province(s) whose name length is longer than 7.

```python
china_provinces = ["GuangDong", "YunNan", "HuBei", "ShanDong", "ZheJiang", "JiangSu", "LiaoNing", "HeiLongJiang"]

# solution 1 using loops
dic_province = {}
for item in china_provinces:
    if len(item) > 7:
        dic_province[item] = len(item)
print (dic_province)

# solution 2 using list comprehension
china_provinces_tuples = [(item, len(item)) for item in china_provinces if len(item)>7]
dic_province = dict(china_provinces_tuples)
print (dic_province)

# solution 3 using dictionary comprehension
dic_province = {item: len(item) for item in china_provinces if len(item)>7}
print (dic_province)
```

### Warnings on Using Dictionary Comprehension
Even though dictionary comprehensions are great for writing elegant code that is easy to read, they are not always the right choice.

We must be careful while using them as :

* They can sometimes make the code run slower and consume more memory.
* They can also decrease the readability of the code.

We must not try to fit a difficult logic or a large number of dictionary comprehension inside them just for the sake of making the code single lined. In these cases, It is better to choose other alternatives like loops.

## 8. Comparison of List and Dictionary
| List | Dictionary |
| ---- | ------- |
| Ordered Sequence of elements | Matching Key and Values |

| Indices has an order | No order is guaranteed |
