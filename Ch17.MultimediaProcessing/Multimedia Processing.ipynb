### Get Intensity Value and Histogram


PIL works with image objects, not files. It automagically closes the file when it doesn't need it any more.

So if you would like to **`convert an image to an array`**, the function *numpy.asarray()* should be called.

```python
# Now we are back to work on PIL/Pillow library
from PIL import Image
import numpy as np
img = Image.open('data/londonbridge.jpeg')
img_arr = np.asarray(img)
print(img_arr.shape)
print(img_arr)
```

PIL/Pillow provides the functions to get the pixel value at a given position or get a single channel of the source image. 

```python
# Return the pixel value at a given position.
print(img.getpixel((200,300)))

# Change the pixel value for given a position.
img.putpixel((200,300),(50, 100, 50))
print(img.getpixel((200,300)))
```

```python
# Return an image containing a single channel of the source image.

# method 1:
print(np.asarray(img.getchannel('R')))
# print(np.asarray(img.getchannel('G')))
# print(np.asarray(img.getchannel('B')))

# method 2:
# or use Image.split() function
r, g, b = img.split() 
print(np.asarray(r))

# method 3:
# or use Image.getdata() function which returns flatted data
r = np.array(img.getdata(band=0))  # get red
g = np.array(img.getdata(band=1))  # get green
b = np.array(img.getdata(band=2))  # get blue
print(np.asarray(r))

# method 4:
# NumPy method
r = np.asarray(img)[:,:,0]  # get red
g = np.asarray(img)[:,:,1]  # get green
b = np.asarray(img)[:,:,2]  # get blue
print(r)
```

Image histogram is a representation of the distribution of intensity values of image pixels. It counts the number of pixels for each (or a range of) tonal value.

```python
# PIL histogram to graph with matplotlib
import matplotlib.pyplot as plt
bins = list(range(256))
r, g, b = img.split() 

plt.plot(bins, r.histogram(), 'r')
plt.plot(bins, g.histogram(), 'g')
plt.plot(bins, b.histogram(), 'b')

plt.xlabel('Pixel value')
plt.ylabel('Frequency')
plt.title("Histogram")
plt.grid(True)
```

### Array Interpolation schemes
[Image interpolation](https://www.cambridgeincolour.com/tutorials/image-interpolation.htm) calculates what the color or value of a pixel "should" be, according to different mathematical schemes. 

One common place that this happens is when you resize an image. The number of pixels change, but you want the same information. Since pixels are discrete, there's missing space. Interpolation is how you fill that space. This is why your images sometimes come out looking pixelated when you blow them up. The effect is more pronounced when the difference between the original image and the expanded image is greater. Let's take our image and shrink it. We're effectively discarding pixels, only keeping a select few. Now when we plot it, that data gets blown up to the size on your screen. The old pixels aren't there anymore, and the computer has to draw in pixels to fill that space.

We'll choose different reampling filters to resize the image.

```python
# nearest: Pick one nearest pixel from the input image. Ignore all other input pixels
img.thumbnail((64, 64), resample = Image.NEAREST) 
imgplot = plt.imshow(img)
```

Let's try some others.

```python
# bilinear: Calculate the output pixel value using linear interpolation on all pixels that may contribute to the output value.
img.thumbnail((64, 64), resample = Image.BILINEAR)  
imgplot = plt.imshow(img)
```

```python
# bicubic: calculate the output pixel value using cubic interpolation on all pixels that may contribute to the output value
img.thumbnail((64, 64), resample = Image.BICUBIC)  # resizes image in-place
imgplot = plt.imshow(img)
```

### Quantize Image
Image.quantize() is to reduce the number of distinct colors used in an image. It [quantize the color through k-means](https://lmcaraig.com/color-quantization-using-k-means).
- method=0: median cut
- method=1: maximum coverage
- method=2: octree
- method=3: libpngquant

```python
img = Image.open('data/garden.jpeg')
display(img)

# colors : The desired number of colors, <= 256
dimg = img.quantize(colors=64)
display(dimg.convert("RGB"))

dimg = img.quantize(colors=32)
display(dimg.convert("RGB"))

dimg = img.quantize(colors=8)
display(dimg.convert("RGB"))
```

### Face Detation
OpenCV provides face detection which is achieved by using Haar cascades. It is a machine learning based approach where a cascade function is trained with a set of input data. OpenCV already contains many pre-trained classifiers for face, eyes, smiles, etc... The pre-trained classifiers should be given beforehand. Normally they can be found in the *opencv/data/haarcascades/* folder, or you can download them from website.

```python
import cv2
from  matplotlib import pyplot as plt

# Load the cascade
face_cascade = cv2.CascadeClassifier('data/haarcascade_frontalface_default.xml')
# Read the input image
img = cv2.imread('data/children.jpeg')
# Convert into grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
# Detect faces, return the coordinates of faces
faces = face_cascade.detectMultiScale(gray, 1.1, 4)
print(faces)
# Draw rectangle around the faces
for (x, y, w, h) in faces:
    cv2.rectangle(img, (x, y), (x+w, y+h), (255, 0, 0), 2)
    
# Display the output
# In OpenCv images are actually represented in BGR order rather than RGB!
# We need to do is convert the image from BGR to RGB:
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.show()
```

## 2.Video Extraction
Given a video, OpenCV provides an easy way to extract frames(images) from the video:
- Open the Video file or camera using cv2.VideoCapture()
- Read frame by frame
- Save each frame using cv2.imwrite()
- Release the VideoCapture and destroy all windows

```python
import cv2
# Opens the Video file
cap= cv2.VideoCapture('data/matching.mov')
success, frame = cap.read()
count=0

# success: True or false refers to whether do we get image or not
while success and count<10:
    cv2.imwrite(str(count)+'.jpg',frame)
    success, frame = cap.read()  
    if success == False:
        break
    count+=1

cap.release()
cv2.destroyAllWindows()
```
