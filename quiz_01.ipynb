```python
from datetime import datetime

form = "%Y-%m-%d"

projects = {
    "a":"2025-06-01",
    "b":"2025-06-01",
    "c":"2024-06-01",
    "d":"2026-06-01",
    "e":"2026-06-01",
}
dic_value = {}
for x,y in projects.items():
    keyNum = datetime.strptime(y, form).year
    
    if dic_value.get(keyNum) is None:
        dic_value[keyNum] = [x]
    else:
        dic_value[keyNum].append(x)

for z in dic_value:
    dic_value[z] = tuple(dic_value[z])

sorted_dict = dict(sorted(dic_value.items()))

print(sorted_dict)
```
